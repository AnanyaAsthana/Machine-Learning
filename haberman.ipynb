```python
import numpy as np
import pandas as pd
import urllib.request
import zipfile
import os

from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

```python
url = "https://sci2s.ugr.es/keel/dataset/data/classification/haberman-5-fold.zip"
zip_name = "haberman-5-fold.zip"

urllib.request.urlretrieve(url, zip_name)
print("Haberman 5-fold dataset downloaded.")
```

```python
with zipfile.ZipFile(zip_name, 'r') as zip_ref:
    zip_ref.extractall("haberman")

print("Dataset extracted.")
```

```python
def load_keel_dat(path):
    data = []
    with open(path, "r") as f:
        for line in f:
            if not line.startswith("@") and line.strip():
                data.append(line.strip().split(","))
    return pd.DataFrame(data)
```

```python
folds = [
    ("haberman/haberman-5-1tra.dat", "haberman/haberman-5-1tst.dat"),
    ("haberman/haberman-5-2tra.dat", "haberman/haberman-5-2tst.dat"),
    ("haberman/haberman-5-3tra.dat", "haberman/haberman-5-3tst.dat"),
    ("haberman/haberman-5-4tra.dat", "haberman/haberman-5-4tst.dat"),
    ("haberman/haberman-5-5tra.dat", "haberman/haberman-5-5tst.dat"),
]
```

```python
k_values = [1, 3, 5, 7, 9]
k_accuracy = {k: [] for k in k_values}

for fold_no, (train_path, test_path) in enumerate(folds, 1):
    print(f"\nFold {fold_no}")

    train_df = load_keel_dat(train_path)
    test_df = load_keel_dat(test_path)

    X_train = train_df.iloc[:, :-1].astype(float)
    y_train = train_df.iloc[:, -1]

    X_test = test_df.iloc[:, :-1].astype(float)
    y_test = test_df.iloc[:, -1]

    for k in k_values:
        model = Pipeline([
            ("scaler", StandardScaler()),
            ("knn", KNeighborsClassifier(n_neighbors=k))
        ])

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        acc = accuracy_score(y_test, y_pred)
        k_accuracy[k].append(acc)

        print(f"  k = {k} → Accuracy = {acc:.4f}")
```

```python
print("\nAverage Accuracy Across 5 Folds:")

best_k = None
best_mean_acc = -1

for k in k_values:
    mean_acc = np.mean(k_accuracy[k])
    print(f"k = {k}, Mean Accuracy = {mean_acc:.4f}")

    if mean_acc > best_mean_acc:
        best_mean_acc = mean_acc
        best_k = k

print("\n✅ Optimal value of k:", best_k)
print("✅ Maximum mean accuracy:", best_mean_acc)
```
